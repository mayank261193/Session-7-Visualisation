### Exercise 9 — Data from an API, then a chart

Call the free **Open-Meteo** API for a 7-day max-temperature forecast for two cities of your choice, build a DataFrame of the average max temp per city, and draw a bar chart.
*(Keep an offline sample so the cell still runs without internet.)*

```python
import requests
import pandas as pd
import matplotlib.pyplot as plt

# TODO: call the API (with an offline fallback), build a DataFrame, and draw a bar chart
```

### Solution

```python
import requests
import pandas as pd
import matplotlib.pyplot as plt

cities = {"Delhi": (28.61, 77.21), "Bengaluru": (12.97, 77.59)}
offline_sample = {"Delhi": [40, 41, 39, 42, 43, 41, 40],
                  "Bengaluru": [31, 32, 30, 31, 33, 32, 31]}

rows = []
for city, (lat, lon) in cities.items():
    url = ("https://api.open-meteo.com/v1/forecast"
           f"?latitude={lat}&longitude={lon}"
           "&daily=temperature_2m_max&timezone=Asia%2FKolkata&forecast_days=7")
    try:
        response = requests.get(url, timeout=10)
        print(city, "status:", response.status_code)
        temps = response.json()["daily"]["temperature_2m_max"]
    except Exception as error:
        print(city, "offline sample (", type(error).__name__, ")")
        temps = offline_sample[city]
    rows.append({"City": city, "AvgMaxTemp_C": round(sum(temps) / len(temps), 1)})

weather = pd.DataFrame(rows)

plt.figure(figsize=(5, 4))
plt.bar(weather["City"], weather["AvgMaxTemp_C"], color="tomato")
plt.title("Average Max Temperature (API)")
plt.ylabel("Temperature (C)")
plt.tight_layout()
plt.show()
```

### Solution

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import matplotlib.pyplot as plt

crop_html = """
<table class="wikitable">
  <tr><th>Crop</th><th>Production_MT</th></tr>
  <tr><td>Rice</td><td>1200</td></tr>
  <tr><td>Wheat</td><td>1050</td></tr>
  <tr><td>Maize</td><td>640</td></tr>
  <tr><td>Pulses</td><td>270</td></tr>
</table>
"""

soup = BeautifulSoup(crop_html, "html.parser")
all_rows = soup.find("table", class_="wikitable").find_all("tr")

headers = [c.get_text(strip=True) for c in all_rows[0].find_all("th")]
data = [[c.get_text(strip=True) for c in row.find_all("td")] for row in all_rows[1:]]

crops = pd.DataFrame(data, columns=headers)
crops["Production_MT"] = crops["Production_MT"].astype(int)

plt.figure(figsize=(6, 4))
plt.bar(crops["Crop"], crops["Production_MT"], color="goldenrod")
plt.title("Crop Production (scraped table)")
plt.ylabel("Production (MT)")
plt.tight_layout()
plt.show()
```
